# Database setup and importing libraries

```python
from common import *
from db import get_engine, setup_database

PROJECT_ROOT = Path.cwd()
```

```python

if not (PROJECT_ROOT / "db.py").is_file():
    PROJECT_ROOT = PROJECT_ROOT.parent

NOTEBOOKS_DIR = PROJECT_ROOT / "notebooks"

# Configure Python import paths
for path in (NOTEBOOKS_DIR, PROJECT_ROOT):
    path = str(path)

    if path in sys.path:
        sys.path.remove(path)

    sys.path.insert(0, path)
```

```python

# ==========================================
# 1. DATABASE INITIALIZATION
# ==========================================

engine = get_engine()

print("=" * 60)
print("CINEMATCH - DATABASE INITIALIZATION")
print("=" * 60)
```

```python

# ==========================================
# 2. DATABASE CONNECTION TEST
# ==========================================

with engine.connect() as conn:
    version = conn.execute(
        text("SELECT version()")
    ).scalar()

    print("\n[OK] PostgreSQL connection established")
    print("PostgreSQL:", version.split(",")[0])

# ==========================================
# 3. DATABASE SETUP
# ==========================================

setup_database()
```

```python

# ==========================================
# 4. MODULE AND PATH VERIFICATION
# ==========================================

print("\n" + "=" * 60)
print("PROJECT CONFIGURATION")
print("=" * 60)

print("Project root:", PROJECT_ROOT)
print("Notebooks directory:", NOTEBOOKS_DIR)
print("Common module:", sys.modules["common"].__file__)
print("Database module:", sys.modules["db"].__file__)

print("\n" + "=" * 60)
print("INITIALIZATION COMPLETED")
print("=" * 60)
```
